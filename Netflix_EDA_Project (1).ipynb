##  Meaningful Questions

Before analysis, we ask the following questions:

- What is the most frequent category/type in the dataset?
- Which country contributes the most content?
- What is the ratio of movies vs TV shows?
- How has content changed over the years?
- Are there any missing or unusual patterns in the data?

```python
df['type'].value_counts()
```

```python
df['country'].value_counts().head(10)
```

```python
df['release_year'].value_counts().sort_index()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(x='type', data=df)
plt.show()
```

```python
df['country'].value_counts().head(10).plot(kind='bar')
plt.show()
```

```python
df['release_year'].value_counts().sort_index().plot()
plt.show()
```




Hypothesis 1:
Movies are more common than TV Shows in the dataset.

Result:
The analysis confirms that movies appear more frequently than TV shows.

Hypothesis 2:
The United States contributes the highest amount of content.

Result:
The country analysis shows that the USA has the highest content count.
